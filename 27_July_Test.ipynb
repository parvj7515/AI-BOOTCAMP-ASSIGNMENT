```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"Roll_No":[101,102,103,104,105,106],"Name":["Parv","Himanshu","Ritesh","Madhav","Bhavya","Saksham"],"Percentage":[92.89,88.00,90.76,87.67,90.90,80.66]}
df=pd.DataFrame(data)
print("Student_Data:")
print(df)
print("\nHighest Marks:",df["Percentage"].max())
plt.bar(df["Name"],df["Percentage"])
plt.title("Result")
plt.xlabel("Students")
plt.ylabel("Percentage")
plt.legend()
plt.show()
```
