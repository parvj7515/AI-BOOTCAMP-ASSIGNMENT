```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
students = pd.read_csv("students.csv")
students
```

```python
students.describe()
```

```python
sorted_students = students.sort_values(
    by="Maths",
    ascending=False
)

sorted_students
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    students["Name"],
    students["Maths"]
)

plt.title("Maths Marks of Students")

plt.xlabel("Student")

plt.ylabel("Marks")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    students["Maths"],
    labels=students["Name"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Maths Marks Distribution")

plt.show()
```
