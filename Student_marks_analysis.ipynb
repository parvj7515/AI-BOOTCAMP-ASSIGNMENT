```python
import pandas as pd
data={"Student":["Aman","Priya","Vijay","Rahul","Aditi"],
      "Maths":[85,95,75,66,89],
      "Science":[99,91,79,86,88],
      "Social_Science":[70,90,69,82,79],
      "Attendance":[95,98,90,96,92]}
student_df=pd.DataFrame(data)
student_df
      
```

```python
student_df.info()
```

```python
student_df.describe()
```

```python
max_marks = student_df['Maths'].max()
min_marks = student_df['Maths'].min()
mean_science = student_df['Science'].mean()
median_attendance = student_df['Attendance'].median()

print(f"Highest Maths Marks: {max_marks}")
print(f"Lowest Maths Marks: {min_marks}")
print(f"Average Science Marks: {mean_science:.2f}")
print(f"Median Attendance: {median_attendance}")
```

```python
sorted_students = student_df.sort_values(by='Maths', ascending=False)

sorted_students
```

```python
top_students = student_df[student_df['Maths'] > 85]

top_students
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(student_df['Student'],
        student_df['Maths'])

plt.title("Maths Marks of Students")

plt.xlabel("Students")

plt.ylabel("Marks")

plt.grid(axis='y')

plt.show()
```
