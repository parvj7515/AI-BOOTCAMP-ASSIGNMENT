```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
employee ={"Name": ["Aman","Riya","Rahul","Priya","Karan","Simran","Aditya","Neha"],
    "Department": ["IT","HR","Finance","IT","Marketing","Finance","HR","IT"],
    "Age": [24,28,30,25,27,29,26,31],
    "Salary": [45000,52000,60000,48000,55000,61000,50000,65000]}
df = pd.DataFrame(employee)
df
```

```python
print("Dataset Information")
df.info()
```

```python
print(df.describe())
```

```python
print("Highest Salary")
print(df["Salary"].max())
```

```python
print("Lowest Salary")
print(df["Salary"].min())
```

```python
print("Average Salary")
print(df["Salary"].mean())
```

```python
highest = df[df["Salary"]==df["Salary"].max()]
highest
```

```python
it = df[df["Department"]=="IT"]
it
```

```python
df.sort_values(by="Salary",ascending=False)
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Name"],df["Salary"])
plt.title("Employee Salary")
plt.xlabel("Employee")
plt.ylabel("Salary")
plt.show()
```

```python
department = df["Department"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(department,labels=department.index,autopct="%1.1f%%")
plt.title("Employees by Department")
plt.show()
```
